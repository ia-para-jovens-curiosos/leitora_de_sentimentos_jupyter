# 💬 IA Leitora de Sentimentos

No jogo **Treine o Robô-Peixe**, vocês treinaram um "robô" mostrando exemplos de peixe e não-peixe. Ele aprendeu a decidir olhando para os exemplos que a turma escolheu — e quando os exemplos eram poucos ou parecidos demais, ele errava.

Agora vamos treinar um robô de verdade, em Python: uma IA que lê uma frase digitada por uma pessoa e adivinha se ela está **positiva** 😀, **negativa** 😞 ou **neutra** 😐.

O "treino" dela vai ser **a lista de palavras que a turma escolher juntos** — ou seja, vocês são os professores dessa IA!

## Como o robô vai aprender

Esse robô não usa internet nem câmera. Ele funciona de um jeito simples, parecido com o que fizemos com as cartas do peixe:

1. A turma monta duas listas de palavras: **palavras positivas** (ex: *ótimo, adorei, feliz*) e **palavras negativas** (ex: *péssimo, odiei, triste*). Essas listas são os *dados de treino*.
2. Quando alguém digita uma frase, o programa quebra a frase em palavras e conta quantas são positivas e quantas são negativas.
3. Quem tiver mais pontos vence: mais palavras positivas → **positivo**; mais negativas → **negativo**; empate ou nenhuma palavra reconhecida → **neutro**.

> ⚠️ Assim como o Robô-Peixe errava quando via pouco exemplo, essa IA só reconhece as palavras que **estiverem nas listas**. Se a turma esquecer uma palavra importante, o robô vai errar — e é exatamente isso que vamos ver e corrigir juntos!

## Passo 1 — Os dados de treino (façam isso em grupo!)

Completem as duas listas abaixo com o máximo de palavras que conseguirem em 10–15 minutos. Podem usar as fichas impressas para organizar as ideias antes de digitar. Escrevam tudo em **letras minúsculas**, sem acento é mais seguro (ex: `otimo` em vez de `ótimo`) — depois explicamos por quê.

```python
# ✍️ PASSO 1: a turma preenche estas duas listas com o máximo de palavras possível.
# Dica: pensem em elogios, reclamações, gírias que vocês usam, emojis escritos por extenso...

palavras_positivas = [
    "otimo", "adorei", "incrivel", "legal", "feliz", "maravilhoso",
    "bom", "amei", "top", "show",
    # ... adicionem mais palavras aqui!
]

palavras_negativas = [
    "pessimo", "odiei", "horrivel", "ruim", "triste", "chato",
    "nojento", "detestei", "lixo", "fraco",
    # ... adicionem mais palavras aqui!
]

print(f"Treino pronto: {len(palavras_positivas)} palavras positivas e {len(palavras_negativas)} negativas.")
```

## Passo 2 — Preparar a frase (limpeza de texto)

Se alguém digitar `"Otimo!!!"` com maiúscula e pontuação, o computador não vai reconhecer isso como a palavra `"otimo"` da nossa lista — para ele são coisas diferentes. Por isso, antes de comparar, a IA precisa **limpar** a frase: deixar tudo minúsculo e tirar a pontuação.

```python
import re

def limpar_frase(frase):
    """Deixa a frase minúscula e remove pontuação, devolvendo a lista de palavras."""
    frase = frase.lower()
    frase = re.sub(r"[^a-zà-ú0-9\s]", "", frase)  # remove pontuação, mantém letras/acentos/números
    return frase.split()

# Teste rápido:
print(limpar_frase("Foi ÓTIMO!!! Eu adorei muito :)"))
```

## Passo 3 — A função que classifica o sentimento

Agora vamos contar quantas palavras da frase aparecem em cada lista de treino, e decidir o resultado.

```python
def classificar_sentimento(frase):
    """Recebe uma frase e devolve ('positivo'/'negativo'/'neutro', pontos_positivos, pontos_negativos)."""
    palavras = limpar_frase(frase)

    pontos_positivos = 0
    pontos_negativos = 0
    for palavra in palavras:
        if palavra in palavras_positivas:
            pontos_positivos += 1
        if palavra in palavras_negativas:
            pontos_negativos += 1

    if pontos_positivos > pontos_negativos:
        resultado = "positivo"
    elif pontos_negativos > pontos_positivos:
        resultado = "negativo"
    else:
        resultado = "neutro"

    return resultado, pontos_positivos, pontos_negativos
```

## Passo 4 — Testando com frases prontas

Antes de deixar o público testar, vamos ver se a IA acerta frases que **nós já sabemos** o sentimento.

```python
frases_teste = [
    "Esse jogo foi incrivel, eu adorei!",
    "Que experiencia horrivel, odiei tudo.",
    "O ceu estava nublado hoje.",
    "Legal mas tambem meio chato.",
]

for frase in frases_teste:
    resultado, pos, neg = classificar_sentimento(frase)
    print(f"{frase!r} -> {resultado} (positivas: {pos}, negativas: {neg})")
```

**Alguma frase saiu errada?** Voltem ao Passo 1, adicionem a palavra que faltou nas listas, executem a célula de novo (Passo 1 → Passo 4) e testem outra vez. Isso *é* treinar a IA — igualzinho ao jogo do Robô-Peixe!

## Passo 5 — Modo estande (para a exibição!)

Esta célula deixa o programa rodando, esperando visitantes digitarem frases. Para sair, digite `sair`.

```python
def rodar_estande():
    print("=" * 50)
    print("  IA LEITORA DE SENTIMENTOS")
    print("  Digite uma frase e descubra o que a IA acha dela.")
    print("  (digite 'sair' para encerrar)")
    print("=" * 50)

    emoji = {"positivo": "😀", "negativo": "😞", "neutro": "😐"}

    while True:
        frase = input("\nDigite sua frase: ")
        if frase.strip().lower() == "sair":
            print("Ate a proxima! 👋")
            break
        resultado, pos, neg = classificar_sentimento(frase)
        print(f"\n{emoji[resultado]}  A IA acha que essa frase e: {resultado.upper()} {emoji[resultado]}")
        print(f"   (encontrei {pos} palavra(s) positiva(s) e {neg} negativa(s) no meu treino)")

rodar_estande()
```

## 🚀 Modo avançado (opcional, se sobrar tempo)

Nossa IA só conhece as palavras que colocamos nas listas — ela é uma IA "caseira". Já existem IAs treinadas com **milhões de frases reais** que entendem muito mais contexto. Vamos comparar?

Essa célula usa a biblioteca `transformers` (Hugging Face) para baixar um modelo pronto e comparar o resultado dele com o nosso robô caseiro. **Isso pode demorar 1–2 minutos** para baixar o modelo na primeira vez, e precisa de internet.

```python
!pip install -q transformers

from transformers import pipeline

analisador_profissional = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

def comparar(frase):
    nosso_resultado, pos, neg = classificar_sentimento(frase)
    profissional = analisador_profissional(frase)[0]
    print(f"Frase: {frase!r}")
    print(f"  Nosso robo caseiro -> {nosso_resultado}")
    print(f"  IA profissional    -> {profissional['label']} (confianca: {profissional['score']:.0%})")

comparar("Esse projeto ficou muito legal, adorei apresentar!")
```

## Para levar para casa

* A IA não "entende" a frase como uma pessoa — ela só reconhece palavras que estavam no treino dela.
* Quanto mais e melhores exemplos de treino, mais ela acerta. Isso vale para essa IA de sentimentos, para o Robô-Peixe, e para praticamente toda IA que existe.
* Um treino incompleto ou tendencioso (por exemplo, só gírias de um grupo) faz a IA errar ou ser injusta com quem fala diferente — é o mesmo problema de **viés** que vimos com as cartas do oceano.
